# Dim Date

```python
def create_date_table(start='2000-01-01', end='2050-12-31'):
  df = pd.DataFrame({"Date": pd.date_range(start, end)})
  df["Year"] = df.Date.dt.year
  df["Month Name"] = df.Date.dt.month_name()
  df["Day Name"] = df.Date.dt.day_name()
  df['Month number'] = df.Date.dt.month
  df['Day of week'] = df.Date.dt.dayofweek
  return df
```

```python
dim_date = create_date_table(start='2023-01-01', end='2023-12-31')
dim_date
```

# Bundled products
*equal or more than 2 items in an order and have products in Finished Products*

## Function BundledProduct()

```python
from collections import defaultdict
from itertools import combinations

def BundledProduct(orders, n=2, min_occurrences=2):
    temp = defaultdict(int)

    for products in orders.values():
        for combo in combinations(products, n):
            sorted_combo = tuple(sorted(combo))
            temp[sorted_combo] += 1

    results = {combo: count for combo, count in temp.items() if count >= min_occurrences}

    return results

# # Example
# # order_id: [order_item_id]
# order = {
#     1: [1, 2],
#     2: [1, 3, 5],
#     3: [1, 5],
#     4: [2, 4],
#     5: [1, 2, 3],
#     6: [1, 2],
#     7: [1, 2, 3, 5, 6],
#     8: [2, 4, 5, 1, 3]
# }
# # d = pd.DataFrame(order.items(), columns=['transaction_id', 'product_id'])
# # # d = pd.Series(order)#.to_frame().reset_index().rename(columns={'index': 'transaction_id', 0: 'product_id'})
# # d
# result = BundledProduct(order, 2, 0)
# print(result)
```

## Orders have 2 or 3 items and items belong to Finished Products

```python
finish_good = order_items[order_items['category_type'] == 'Finished Products']
finish_good_condition = finish_good[finish_good['transaction_id'].isin(finish_good.groupby('transaction_id').
                                   agg(rows = ('product_id', 'nunique')).
                                   reset_index()[finish_good.groupby('transaction_id').
                                                 agg(rows = ('product_id', 'nunique')).
                                                 reset_index()['rows'] >= 3]['transaction_id']
                                   )
][['transaction_id', 'product_id']].drop_duplicates()
finish_good_condition
```

```python
# change product_id in transaction_id to list
product_list = finish_good_condition.groupby('transaction_id')['product_id'].apply(list).reset_index()
product_list
```

```python
product_list_dict = product_list.set_index('transaction_id')['product_id'].to_dict()
result_func_3 = BundledProduct(product_list_dict, 3, 0)
result_func_3 = pd.DataFrame(result_func_3.items(), columns=['product_id', 'count'])
result_func_3 = result_func_3.sort_values('count', ascending=False)
result_func_3
```

```python
result_func_2 = BundledProduct(product_list_dict, 2, 0)
result_func_2 = pd.DataFrame(result_func_2.items(), columns=['product_id', 'count'])
result_func_2 = result_func_2.sort_values('count', ascending=False)
result_func_2
```
